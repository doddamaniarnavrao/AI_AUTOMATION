```python
hero = input("Who do you want to talk to? ")
place = input("Where are they right now? ")
mood = input("What mood are they in? (happy, grumpy, sleepy...) ")
print(f"Opening a portal to {place}... connecting to {hero}, who feels {mood}!")
```

```python
# Python fills in the blanks to write the AI's secret rules
system_prompt = f"""You are {hero}. You are in {place} and you feel {mood}.
Stay in character at all times.
Keep every answer under 3 sentences.
Be kind and suitable for a 9-year-old."""
print("--- The AI's secret rules ---")
print(system_prompt)
```

```python
print(f"PORTAL OPEN! You are talking to {hero}. Type quit to close it.")
while True:
  message = input("Arnav: ")
  if message == "quit":
    print("Portal closed. See you in another universe!")
    break
# Send the secret rules + Arnav's message to the AI brain
  response = client.models.generate_content(
    model="gemini-3.8-flash",
    contents=system_prompt + "\nArnav says: " + message
)
print(hero + ":", response.text)
```
