```python
import os
import subprocess
import sys


def ensure_google_genai():
    try:
        from google import genai  # noqa: F401
        return
    except ImportError:
        print("Installing google-genai...")
        subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "google-genai"])


ensure_google_genai()

from google import genai


def get_api_key():
    if "google.colab" in sys.modules:
        try:
            from google.colab import userdata

            key = userdata.get("GOOGLE_API_KEY")
            if key:
                return key
        except Exception:
            pass

    key = os.getenv("GOOGLE_API_KEY")
    if key:
        return key

    if sys.stdin.isatty():
        return input("Paste your Google API key (or set GOOGLE_API_KEY): ").strip()

    return ""


def build_system_prompt(hero, place, mood):
    return (
        f"You are {hero}. You are in {place} and you feel {mood}.\n"
        "Stay in character at all times.\n"
        "Keep every answer under 3 sentences.\n"
        "Be kind and suitable for a 9-year-old."
    )


def main():
    print("Checking environment: VS Code or Colab")
    api_key = get_api_key()
    if not api_key:
        raise SystemExit("No Google API key found. Set GOOGLE_API_KEY or paste one when prompted.")

    client = genai.Client(api_key=api_key)
    print("Portal power: ONLINE (Client initialized)")

    hero = "Falcon"
    place = "XYG"
    power_level = 6075555575646754674567456755595
    print(hero, "lives in", place)
    print(f"{hero} lives in {place} and has power level {power_level}!")
    print(f"After training, power level is {power_level * 90}!")

    hero = input("Who do you want to talk to? ").strip() or hero
    place = input("Where are they right now? ").strip() or place
    mood = input("What mood are they in? (happy, grumpy, sleepy...) ").strip() or "happy"
    print(f"Opening a portal to {place}... connecting to {hero}, who feels {mood}!")

    system_prompt = build_system_prompt(hero, place, mood)
    print("--- The AI's secret rules ---")
    print(system_prompt)

    print(f"PORTAL OPEN! You are talking to {hero}. Type 'quit' to close it.")

    while True:
        message = input("Arnav: ").strip()
        if message.lower() == "quit":
            print("Portal closed. See you in another universe!")
            break

        response = client.models.generate_content(
            model="gemini-2.0-flash",
            contents=f"{system_prompt}\nArnav says: {message}",
        )
        print(f"{hero}: {response.text}")


if __name__ == "__main__":
    main()
```
