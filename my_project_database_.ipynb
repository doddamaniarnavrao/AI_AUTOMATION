###cell 1: build the database


```python
import pandas as pd
projects = pd.DataFrame({
"project": ["Maze Runner", "Chat Bot", "Flappy Bird", "Calculator App",
"Voice Helper", "Space Shooter", "Quiz App", "Story Teller AI",
"Dancing Cat", "Snake Game", "Weather App", "Persona Bot"],
"course":  ["Pygame", "Python AI", "Pygame", "App Inventor",
"Voiceflow", "Pygame", "App Inventor", "Python AI",
"Scratch", "Pygame", "App Inventor", "Python AI"],
"type":
["Game", "AI", "Game", "App", "AI", "Game",
"App", "AI", "Art", "Game", "App", "AI"],
"stars":   [4, 5, 5, 3, 4, 5, 3, 5, 4, 4, 3, 5],
})
print(projects)
print("Total projects:", len(projects))
```

### step 2: filtering only ai agents



```python
ai_projects = projects[projects["type"] == "AI"]
print(ai_projects)
print("AI projects:", len(ai_projects))
```

CELL 3 · Sorting: my top 3 favourites

```python
favourites = projects.sort_values("stars", ascending=False)
print(favourites.head(3))
```

CELL 4 · Group By: which course had the most projects?

```python
per_course = projects.groupby("course").size().sort_values(ascending=False)
print(per_course)
print("Most projects:", per_course.index[0])
```

GAME · The Database Quiz

```python
score = 0
guess = int(input("How many GAME projects are there? "))
answer = len(projects[projects["type"] == "Game"])
if guess == answer:
  print("Correct!")
  score += 1
else:
   print("The database says", answer)

guess = int(input("How many projects got 5 stars? "))
answer = len(projects[projects["stars"] == 5])
if guess == answer:
  print("Correct!")
  score += 1
else:
  print("The database says", answer)
print(f"Quiz score: {score} out of 2")
```
