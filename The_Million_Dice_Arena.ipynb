```python
arnav = np.random.randint(1, 7, size=5)
computer = np.random.randint(1, 7, size=5)

arnav = np.random.randint(1, 7, size=5)
computer = np.random.randint(1, 7, size=5)

arnav_wins = np.sum(arnav > computer)
computer_wins = np.sum(computer > arnav)
print(f"Arnav won {arnav_wins} rounds. Computer won {computer_wins} rounds.")

if arnav_wins > computer_wins:
  print("ARNAV IS THE DICE CHAMPION!")
elif computer_wins > arnav_wins:
  print("The computer wins this time. Re-run for a rematch!")
else:
  print("It's a TIE!")
```

```python
rolls = np.random.randint(1, 7, size=100_000)

print("How many dice?", len(rolls))
print("First 10 rolls:", rolls[:1000])

for number in range(0, 9):
     count = np.sum(rolls == number)
     print(f"{number} came up {count} times")
```

```python
import random
import time

# Racer 1: a normal loop rolls one dice at a time
start = time.time()
slow_rolls = []
for i in range(1_000_000):
    slow_rolls.append(random.randint(1, 6))
loop_time = time.time() - start

# Racer 2: NumPy rolls them all at once
start = time.time()
fast_rolls = np.random.randint(1, 7, size=1_000_000)
numpy_time = time.time() - start

print(f"Loop took  {loop_time:.3f} seconds")
print(f"NumPy took {numpy_time:.3f} seconds")
print(f"NumPy was {loop_time / numpy_time:.0f} times faster!")
```

```python
guess = int(input("Which total (2 to 12) will appear the MOST? "))

die1 = np.random.randint(1, 7, size=1_000_000)
die2 = np.random.randint(1, 7, size=1_000_000)
totals = die1 + die2        # adds a million pairs in one go!

best_total = 0
best_count = 0
for total in range(2, 13):
     count = np.sum(totals == total)
     print(f"Total {total}: " + "#" * (count // 5000))
     if count > best_count:
        best_count = count
        best_total = total

print(f"The winner is {best_total}!")
if guess == best_total:
     print("BOSS DEFEATED! You predicted it!")
else:
    print("The boss wins this time. Can you work out WHY it won?")
```
